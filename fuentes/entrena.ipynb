```python
import mlflow

# Creamos o seleccionamos el experimento para tu tarea
mlflow.set_experiment("Actividad5_Modelos")
```

```python
with mlflow.start_run(run_name="Prueba_Inicial"):
    # 1. Parámetros
    mlflow.log_param("algoritmo", "Regresion Logistica")
    mlflow.log_param("regularizacion", 0.01)

    # 2. Métricas
    mlflow.log_metric("accuracy", 0.88)
    mlflow.log_metric("f1_score", 0.85)

    # 3. Artefactos
    with open("reporte_prueba.txt", "w") as f:
        f.write("Prueba de configuracion exitosa para la Fase 2.")

    mlflow.log_artifact("reporte_prueba.txt")

print("¡MLflow configurado y probado con éxito!")
```
